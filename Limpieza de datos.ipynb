```python
cat_cols = ["Marital status", "Application mode", "Course", "Previous qualification",
            "Nacionality", "Mother's qualification", "Father's qualification",
            "Mother's occupation", "Father's occupation"]
df[cat_cols] = df[cat_cols].astype("category")

num_cols = [c for c in df.select_dtypes("number").columns
            if c not in cat_cols and df[c].nunique() > 2]
```

```python
def grupo_otros(columna, min_frec=0.01, valor_otro=-1):
    columna = columna.astype(int)
    freq = columna.value_counts(normalize=True)
    rare = freq[freq < min_frec].index
    return columna.where(~columna.isin(rare), valor_otro).astype("category")

for col in cat_cols:
    antes = df[col].nunique()
    df[col] = grupo_otros(df[col])
    print(col, antes, "->", df[col].nunique())
```

```python
for periodo in ["1st", "2nd"]:
    p = f"Curricular units {periodo} sem ("
    enr = df[p + "enrolled)"]
    ev  = df[p + "evaluations)"]
    apr = df[p + "approved)"]

    df[f"estado_{periodo}"] = np.select(
        [enr == 0, (enr > 0) & (ev == 0), (ev > 0) & (apr == 0)],
        ["no_inscrito", "inscrito_sin_eval", "evaluado_sin_aprobar"],
        default="normal",
    )

print(df["estado_1st"].value_counts())
print(df["estado_2nd"].value_counts())
```

```python
print(df["Age at enrollment"].quantile([0.01, 0.05, 0.25, 0.50, 0.75, 0.95, 0.99]))
df["es_adulto"] = (df["Age at enrollment"] >= 25).astype(int)
```

```python
umbral = 0.95

proporcion_moda = df.apply(
    lambda columna: columna.value_counts(normalize=True, dropna=False).iloc[0]
)

columnas_mas_95 = proporcion_moda[proporcion_moda >= umbral].sort_values(
    ascending=False
)

print("Columnas con más del 95% del mismo valor:")
print(columnas_mas_95)
```

```python
df = df.drop(columns=columnas_mas_95.index[0])
```

```python
filas = []
for c in num_cols:
    s = df[c]
    q1, q3 = s.quantile([0.25, 0.75])
    iqr = q3 - q1
    lo, hi = q1 - 1.5 * iqr, q3 + 1.5 * iqr
    n_out = ((s < lo) | (s > hi)).sum()
    p99 = s.quantile(0.99)
    filas.append({
        "variable": c,
        "mediana": s.median(),
        "p99": p99,
        "max": s.max(),
        "max/mediana": s.max() / s.median() if s.median() > 0 else np.nan,
        "skew": s.skew(),
        "% ceros": (s == 0).mean() * 100,
        "outliers_IQR": n_out,
        "% outliers_IQR": n_out / len(s) * 100,
        "sobre_p99": (s > p99).sum(),
    })

resumen = pd.DataFrame(filas).set_index("variable").round(2)
resumen = resumen.sort_values("% outliers_IQR", ascending=False)
print(resumen.to_string())
```

```python
P = lambda sem, col: f"Curricular units {sem} sem ({col})"

for s in ["1st", "2nd"]:
    inscritas = df[P(s, "enrolled")]
    aprobadas = df[P(s, "approved")]
    # Si no inscribió nada, la tasa queda en 0 (ese caso ya lo marca estado_1st/2nd)
    df[f"tasa_aprob_{s}"] = np.where(inscritas > 0, aprobadas / inscritas, 0)

print(df[["tasa_aprob_1st", "tasa_aprob_2nd"]].describe().round(2))
print("NaN:", df[["tasa_aprob_1st", "tasa_aprob_2nd"]].isna().sum().sum())
```
